```python
df['Product_Category']=df['Product_Category'].replace({'Electrnc':'Electronics' , 'clothing' :'Clothing'})
category_sales=df.groupby('Product_Category')['Total_Amount_Formula'].sum().reset_index()
category_sales=category_sales.sort_values(by='Total_Amount_Formula' , ascending=False)
print(category_sales)
```

```python
pay_method=df['Payment_Method'].value_counts().reset_index()
print(pay_method)
```

```python
returns_no=df['Return_Status'].value_counts().reset_index()
print(returns_no)
```

```python
most_customer_buy=df['Customer_Name'].value_counts().reset_index()
print(most_customer_buy)
```

```python
df['order_month']=df['Date'].dt.to_period('M')
print(df['order_month'])
```

```python
monthly_sales=df.groupby('order_month').agg(month_sales=('Total_Amount_Formula','sum')).reset_index()
monthly_sales['order_month']=monthly_sales['order_month'].astype(str)
print(monthly_sales)                                         
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=category_sales,x='Product_Category' , y='Total_Amount_Formula', hue ='Product_Category' ,palette='viridis' , legend=False) 
plt.title('total sales by product category',fontsize=14 , fontweight='bold')
plt.xlabel('Product_Category',fontsize=12)
plt.ylabel('Total_Amount_Formula',fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=monthly_sales,x='order_month' , y='month_sales', hue ='order_month' ,palette='viridis' , legend=False) 
plt.title('Monthly Sales',fontsize=14 , fontweight='bold')
plt.xlabel('Order Month',fontsize=12)
plt.ylabel('Month Sales',fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_monthly_sales=df.groupby(['order_month','Product_Category'])['Total_Amount_Formula'].sum().reset_index()
print(category_monthly_sales)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=category_monthly_sales, x='order_month' , y='Total_Amount_Formula', hue ='Product_Category' ,palette='viridis' ) 
plt.title('Sales by Category Per Month',fontsize=14 , fontweight='bold')
plt.xlabel('Order Month',fontsize=12)
plt.ylabel('Total Sales',fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Product Category')
plt.tight_layout()
plt.show()
```
